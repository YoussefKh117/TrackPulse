```python
from pyspark.sql import functions as F

# 1. Read trips.json correctly (JSON array)
trips_df = (
    spark.read
    .option("multiLine", True)
    .json(trips_path)
)

print("=== TRIPS SCHEMA ===")
trips_df.printSchema()
display(trips_df.limit(5))

print("Trips rows:", trips_df.count())


# 2. Inspect anomaly labels
anomaly_path = historical_path + "anomaly_labels/"
anomaly_files = dbutils.fs.ls(anomaly_path)

print("Number of anomaly files:", len(anomaly_files))
display(anomaly_files[:3])

# Read only the first anomaly file for now
first_anomaly_file = anomaly_files[0].path

anomalies_df = spark.read.json(first_anomaly_file)

print("=== ANOMALY LABELS SCHEMA ===")
anomalies_df.printSchema()
display(anomalies_df.limit(5))


# 3. Check master tables and duplicate IDs
master_keys = {
    "drivers.csv": "driver_id",
    "routes.csv": "route_id",
    "shipments.csv": "shipment_id",
    "trip_plans.csv": "trip_id",
    "vehicles.csv": "vehicle_id"
}

for file_name, key in master_keys.items():
    df = (
        spark.read
        .option("header", True)
        .option("inferSchema", True)
        .csv(master_path + file_name)
    )

    total_rows = df.count()

    null_keys = df.filter(
        F.col(key).isNull() | (F.trim(F.col(key)) == "")
    ).count()

    duplicate_keys = (
        df.groupBy(key)
        .count()
        .filter(F.col("count") > 1)
        .count()
    )

    print(
        f"{file_name}: rows={total_rows}, "
        f"null_keys={null_keys}, "
        f"duplicate_key_groups={duplicate_keys}"
    )
```

```python
from pyspark.sql import functions as F

# 1. Profile a single telemetry chunk
print("=== TELEMETRY QUALITY ===")

telemetry_quality = (
    sample_df.agg(
        F.count("*").alias("total_rows"),
        F.countDistinct("event_id").alias("unique_event_ids"),
        F.sum(F.col("event_id").isNull().cast("int")).alias("null_event_ids"),
        F.sum(F.col("vehicle_id").isNull().cast("int")).alias("null_vehicle_ids"),
        F.sum(F.col("trip_id").isNull().cast("int")).alias("null_trip_ids"),
        F.sum(F.col("event_timestamp").isNull().cast("int")).alias("null_timestamps"),
        F.sum(F.col("latitude").isNull().cast("int")).alias("null_latitudes"),
        F.sum(F.col("longitude").isNull().cast("int")).alias("null_longitudes")
    )
)

display(telemetry_quality)


# 2. Check numeric ranges on the same sample
display(
    sample_df.select(
        F.min("speed_kph").alias("min_speed"),
        F.max("speed_kph").alias("max_speed"),
        F.min("engine_temp_c").alias("min_engine_temp"),
        F.max("engine_temp_c").alias("max_engine_temp"),
        F.min("latitude").alias("min_latitude"),
        F.max("latitude").alias("max_latitude"),
        F.min("longitude").alias("min_longitude"),
        F.max("longitude").alias("max_longitude")
    )
)


# 3. Check nulls in trips
print("=== TRIPS QUALITY ===")

display(
    trips_df.agg(
        F.count("*").alias("total_trips"),
        F.countDistinct("trip_id").alias("unique_trip_ids"),
        F.sum(F.col("actual_end").isNull().cast("int")).alias("null_actual_end"),
        F.sum(F.col("distance_km").isNull().cast("int")).alias("null_distance"),
        F.sum(F.col("fuel_used_liters").isNull().cast("int")).alias("null_fuel")
    )
)


# 4. Check anomaly labels in the first chunk
print("=== ANOMALY LABELS QUALITY ===")

display(
    anomalies_df.groupBy("anomaly_type", "severity")
    .count()
    .orderBy(F.desc("count"))
)


# 5. Check whether trips and trip plans agree on IDs
plans_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(master_path + "trip_plans.csv")
)

print("=== TRIPS VS TRIP PLANS ===")

display(
    trips_df.alias("t")
    .join(
        plans_df.select("trip_id", "planned_start", "planned_end", "trip_status").alias("p"),
        on="trip_id",
        how="inner"
    )
    .select(
        "trip_id",
        F.col("t.planned_start").alias("trips_planned_start"),
        F.col("p.planned_start").alias("plans_planned_start"),
        F.col("t.planned_end").alias("trips_planned_end"),
        F.col("p.planned_end").alias("plans_planned_end"),
        F.col("t.trip_status").alias("trips_status"),
        F.col("p.trip_status").alias("plans_status")
    )
    .limit(10)
)
```

```python
from pyspark.sql import functions as F

# 1. Check anomaly labels against the first telemetry chunk
print("=== ANOMALY LABEL MATCHING ===")

print("Telemetry rows:", sample_df.count())
print("Anomaly label rows:", anomalies_df.count())

print(
    "Unique anomaly event IDs:",
    anomalies_df.select("event_id").distinct().count()
)

# How many anomaly records match a telemetry event?
matched_anomalies = anomalies_df.join(
    sample_df.select("event_id"),
    on="event_id",
    how="inner"
)

unmatched_anomalies = anomalies_df.join(
    sample_df.select("event_id"),
    on="event_id",
    how="left_anti"
)

print("Matched anomaly records:", matched_anomalies.count())
print("Unmatched anomaly records:", unmatched_anomalies.count())

# Check if any event has multiple anomaly-label records
display(
    anomalies_df.groupBy("event_id")
    .count()
    .filter(F.col("count") > 1)
    .limit(10)
)


# 2. Compare all trips with trip plans
print("=== TRIP STATUS COMPARISON ===")

trip_comparison = (
    trips_df.alias("t")
    .join(
        plans_df.alias("p"),
        on="trip_id",
        how="inner"
    )
)

display(
    trip_comparison.groupBy(
        F.col("t.trip_status").alias("trips_status"),
        F.col("p.trip_status").alias("plans_status")
    )
    .count()
)

# Compare planned end times in hours
display(
    trip_comparison.select(
        "trip_id",
        F.col("t.planned_end").alias("trips_planned_end"),
        F.col("p.planned_end").alias("plans_planned_end"),
        (
            F.unix_timestamp(
                F.to_timestamp("t.planned_end")
            )
            - F.unix_timestamp(
                F.to_timestamp("p.planned_end")
            )
        ).alias("planned_end_difference_seconds")
    )
    .groupBy("planned_end_difference_seconds")
    .count()
    .orderBy(F.desc("count"))
)


# 3. Check routes numeric columns before cleaning
print("=== ROUTE DATA TYPES ===")

routes_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(master_path + "routes.csv")
)

display(
    routes_df.select(
        "route_id",
        "planned_distance_km",
        "planned_duration_min"
    )
)
```
